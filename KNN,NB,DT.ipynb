```python
path="/content/drive/MyDrive/disease_ml_dataset.csv"
```

```python
import pandas as pd
df = pd.read_csv(path)
```

```python
df.head()
```

```python
df['Chest_Pain'] = df["Chest_Pain"].replace({
   "No": 0,
   "Yes": 1
})
```

```python
df['Disease'] = df['Disease'].replace({
    "Yes":1,
    "No": 0
})
```

```python
df.head()
```

```python
x=df.drop("Disease",axis=1)
y=df["Disease"]
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
num_col=["Age","Oxygen_Level","Heart_Rate","WBC_Count"]
scaler = StandardScaler()
```

```python
X_train[num_col] = scaler.fit_transform(X_train[num_col])
X_test[num_col] = scaler.transform(X_test[num_col])
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)
```

```python
KNN = KNeighborsClassifier(n_neighbors=5)
KNN.fit(X_train, y_train)
```

```python
KNN_PRED = KNN.predict(X_test)
```

```python
KNN_PRED
```

```python
print("Actual:", y_test.values)
print("Predicted:", KNN_PRED)
```

```python
cn = confusion_matrix(y_test, KNN_PRED)
print(cn)
```

```python
accuracy = accuracy_score(y_test, KNN_PRED)
print(accuracy)
```

```python
print(classification_report(y_test, KNN_PRED))
```

```python
from sklearn.metrics import (


    precision_score,
    recall_score,
    f1_score,

)
```

```python
print("Accuracy :", accuracy_score(y_test, KNN_PRED))
print("Precision:", precision_score(y_test, KNN_PRED))
print("Recall   :", recall_score(y_test, KNN_PRED))
print("F1 Score :", f1_score(y_test, KNN_PRED))
```

```python
from sklearn.naive_bayes import GaussianNB
NB = GaussianNB()
NB.fit(X_train, y_train)
```

```python
NB_pred = NB.predict(X_test)
```

```python
NB_pred
```

```python
print("Actual:", y_test.values)
print("Predicted:", NB_pred)
```

```python
cn = confusion_matrix(y_test, NB_pred)
print(cn)
```

```python
print(classification_report(y_test, NB_pred))
```

```python
print("Accuracy :", accuracy_score(y_test, NB_pred))
print("Precision:", precision_score(y_test, NB_pred))
print("Recall   :", recall_score(y_test, NB_pred))
print("F1 Score :", f1_score(y_test, NB_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
DT = DecisionTreeClassifier()
DT.fit(X_train, y_train)
```

```python
DT_pred = DT.predict(X_test)
DT_pred
```

```python
print("Actual:", y_test.values)
print("Predicted:", DT_pred)
```

```python
cn = confusion_matrix(y_test, DT_pred)
print(cn)
```

```python
print("Accuracy :", accuracy_score(y_test, DT_pred))
print("Precision:", precision_score(y_test, DT_pred))
print("Recall   :", recall_score(y_test, DT_pred))
print("F1 Score :", f1_score(y_test, DT_pred))
```

```python
knn_accuracy = accuracy_score(y_test, KNN_PRED)
nb_accuracy = accuracy_score(y_test, NB_pred)
dt_accuracy = accuracy_score(y_test, DT_pred)

print("KNN:", knn_accuracy)
print("Naive Bayes:", nb_accuracy)
print("Decision Tree:", dt_accuracy)
```

```python
results = pd.DataFrame({
    "Model": ["KNN", "Naive Bayes", "Decision Tree"],
    "Accuracy": [
        knn_accuracy,
        nb_accuracy,
        dt_accuracy
    ]
})

print(results)
```
