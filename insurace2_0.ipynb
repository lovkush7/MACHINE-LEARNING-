```python
def plotting(var, num):
  plt.subplot(3,2,num)
  sns.histplot(data[var],kde=True)

plotting('age',1)
plotting('bmi',2)
plotting('children',3)
plotting('annual_income',4)
plotting('medical_history_score',5)
plotting('insurance_charge',6)
plt.tight_layout()
```

```python
numeric_col = ['age','bmi','children','annual_income','medical_history_score','insurance_charge']

for col in numeric_col:
  plt.figure(figsize=(9,6))
  sns.histplot(data[col],kde=True, bins=20)
```

```python
for col in numeric_col:

    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[col] < lower) |
        (data[col] > upper)
    ]

    print(col)
    print("Lower:", lower)
    print("Upper:", upper)
    print("Outliers:", len(outliers))
    print()
```
